# Run retrieval experiments

Run the configured retrieval models across all datasets and compare the saved metrics.

To add a future variation, register it in `retrieval/factory.py`, add its settings to the config if needed, and add its name to `MODELS` below.

```python
from pathlib import Path
import json
import sys

import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

SRC_DIR = PROJECT_ROOT / "src"
SCRIPTS_DIR = PROJECT_ROOT / "scripts"

sys.path.insert(0, str(SRC_DIR))
sys.path.insert(0, str(SCRIPTS_DIR))

from evaluate import main
```

## Choose models

The names must match entries registered in `RETRIEVER_BUILDERS`.

```python
MODELS = [
    "bm25",
    "dense",
    # "dense_finetuned",
    # "dense_hard_negative",
    # "dense_pseudo_label",
    # "hybrid",
]
```

## Run experiments

```python
main(models=MODELS)
```

## Compare results

```python
rows = []

for model_name in MODELS:
    result_path = PROJECT_ROOT / "reports" / "results" / f"{model_name}.json"

    with open(result_path) as f:
        result = json.load(f)

    for dataset_name, dataset_result in result["datasets"].items():
        rows.append({
            "model": model_name,
            "dataset": dataset_name,
            **dataset_result["metrics"],
        })

results_df = pd.DataFrame(rows)
results_df
```
